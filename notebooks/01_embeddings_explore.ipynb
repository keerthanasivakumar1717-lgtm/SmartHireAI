# SmartHireAI: 01. Embeddings Exploration & Similarity Analysis
This notebook demonstrates semantic embedding generation and cosine similarity comparisons between candidate profiles and job descriptions.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import numpy as np
from src.search.embed import generate_embedding, generate_embeddings_batch
```

### 1. Generate Embeddings for Sample Job Titles and Candidate Profiles

```python
sample_texts = [
    'Junior Data Analyst with Python and SQL experience',
    'Backend Software Engineer specializing in FastAPI and PostgreSQL',
    'Frontend React and TypeScript UI Developer',
    'Machine Learning Engineer with PyTorch and NLP backgrounds'
]

embeddings = generate_embeddings_batch(sample_texts)
print('Embeddings shape:', embeddings.shape)
print('Embedding dimension (D):', embeddings.shape[1])
```

### 2. Compute Cosine Similarity Matrix

```python
sim_matrix = np.dot(embeddings, embeddings.T)
print('Cosine Similarity Matrix (Texts 0 to 3):')
for i, row in enumerate(sim_matrix):
    print(f'Text {i} similarities:', [round(float(s), 3) for s in row])
```

### 3. Test Candidate Query Matching

```python
candidate_query = 'Candidate proficient in SQL queries, data cleaning, and Tableau dashboards'
cand_emb = generate_embedding(candidate_query)

scores = np.dot(embeddings, cand_emb)
for text, score in zip(sample_texts, scores):
    norm_score = max(0.0, min(1.0, (score + 1.0) / 2.0))
    print(f'Match: {round(norm_score * 100, 1)}% | {text}')
```
