# SmartHireAI: 02. Build FAISS Vector Index
This notebook loads the job postings dataset, normalizes columns, embeds descriptions, and builds a persisted FAISS index.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

from src import config
from src.search.job_search import JobSearchIndex
```

### 1. Initialize and Build Index from CSV

```python
searcher = JobSearchIndex()
print('Building index from:', config.DEFAULT_JOBS_CSV)
count = searcher.build_index(config.DEFAULT_JOBS_CSV)
print(f'Successfully indexed {count} job descriptions!')
```

### 2. Verify Index Files on Disk

```python
print('Vectorstore contents:')
for f in config.VECTORSTORE_DIR.iterdir():
    print(' -', f.name, f'({f.stat().st_size} bytes)')
```

### 3. Run Test Top-N Semantic Query

```python
results = searcher.search('Looking for Python backend developer with FastAPI and Docker', top_k=3)
for i, res in enumerate(results, 1):
    print(f'[{i}] {res["title"]} @ {res["company"]} ({res["match_percentage"]}%)')
    print('    Matched skills:', res['matched_skills'])
    print('    Missing skills:', res['missing_skills'])
```
